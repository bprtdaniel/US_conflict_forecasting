### 1. Generate function to extract data

```python
# create empty df for codebook
codebook = pd.DataFrame(columns = ['var_name', 'var_label', 'var_encoding', 'var_level', 'var_length', 'data_source'])
```

```python
# define function to load data and extract vars
def extract_vars(file_path, codebook):
    # load data
    data = pd.read_csv(file_path)
    
    # extract file name
    file_name = os.path.basename(file_path)
    
    # extract column names
    columns = data.columns.tolist()

    # filter columns for unique var names
    unique_columns = [col for col in columns if col not in codebook['var_name'].unique()]

    # calculate variable length for each unique column
    variable_lengths = []
    for col in unique_columns:
        variable_lengths.append(data[col].dropna().unique().size)
    
    # create temporary dataset
    df_tmp = pd.DataFrame(columns = ['var_name', 'data_source'])
    df_tmp['var_name'] = unique_columns
    df_tmp['var_length'] = variable_lengths
    df_tmp['data_source'] = file_name

    # append to existing codebook
    return pd.concat([codebook, df_tmp], ignore_index = True)
```

### 2. Apply function to datasets

```python
# directory containing datasets
directory_path = '/Users/varvarailyina/hertie/ML/US_conflict_forecasting/2_Dataset/b_Partial_Datasets'

# list all files in directory
file_names = os.listdir(directory_path)

# filter for .csv files
csv_files = [file for file in file_names if file.endswith('.csv')]

# run function over each .csv file
for file_name in csv_files:
    full_path = os.path.join(directory_path, file_name)
    codebook = extract_vars(full_path, codebook)
```
