```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Loading the datasets
data_2022 = pd.read_csv('/content/PA_l2_2022stats_county.csv')
data_2020 = pd.read_csv('/content/PA_l2_2022stats_2020block.csv')

# Displaying the first 5 rows of the dataset
display(data_2022.head())
display(data_2020.head())
```

```python
# Printing the column names
print(data_2022.columns)
print(data_2020.columns)
```

```python
# Checking for missing values
data_2022.dropna(inplace=True)
data_2020.dropna(inplace=True)

display(data_2022.head())
display(data_2020.head())
```

```python
# Finding the maximum votes received by each party in each row
data_2022['max_votes'] = data_2022[['party_dem', 'party_rep', 'party_npp', 'party_rin', 'party_lib', 'party_grn', 'party_con', 'party_ind', 'party_oth', 'party_unk']].max(axis=1)

# Calculating the win margins for each party as the difference from the max votes
data_2022['margin_dem'] = data_2022['max_votes'] - data_2022['party_dem']
data_2022['margin_rep'] = data_2022['max_votes'] - data_2022['party_rep']
data_2022['margin_npp'] = data_2022['max_votes'] - data_2022['party_npp']
data_2022['margin_rin'] = data_2022['max_votes'] - data_2022['party_rin']
data_2022['margin_lib'] = data_2022['max_votes'] - data_2022['party_lib']
data_2022['margin_grn'] = data_2022['max_votes'] - data_2022['party_grn']
data_2022['margin_con'] = data_2022['max_votes'] - data_2022['party_con']
data_2022['margin_ind'] = data_2022['max_votes'] - data_2022['party_ind']
data_2022['margin_oth'] = data_2022['max_votes'] - data_2022['party_oth']
data_2022['margin_unk'] = data_2022['max_votes'] - data_2022['party_unk']

# Displaying the data with win margins
display(data_2022[['countyname', 'margin_dem', 'margin_rep', 'margin_npp', 'margin_rin', 'margin_lib', 'margin_grn', 'margin_con', 'margin_ind', 'margin_oth', 'margin_unk']])
```

```python
# Saving DataFrame with win margins to a CSV file
data_2022[['countyname', 'margin_dem', 'margin_rep', 'margin_npp', 'margin_rin', 'margin_lib', 'margin_grn', 'margin_con', 'margin_ind', 'margin_oth', 'margin_unk']].to_csv('County_Margins_2022.csv', index=False)
```

EXPLANATION:
Where the party shows a margin of '0', it means that the party won the elections in that county and the others with high numbers is the difference of votes they have with the winning party with '0'

EX: For county: BERKS, the republicans won the elections in that county and the democrats had 720 fewer votes.

```python
# Aggregating the sum of votes for each party in 2020 and 2022
sum_2020 = data_2020[[
    'party_dem', 'party_rep', 'party_npp', 'party_rin',
    'party_lib', 'party_grn', 'party_con', 'party_ind',
    'party_oth', 'party_unk'
]].sum()

sum_2022 = data_2022[[
    'party_dem', 'party_rep', 'party_npp', 'party_rin',
    'party_lib', 'party_grn', 'party_con', 'party_ind',
    'party_oth', 'party_unk'
]].sum()

# Creating a DataFrame to display the results
results_df = pd.DataFrame({
    'Party': sum_2020.index,
    'Votes_2020': sum_2020.values,
    'Votes_2022': sum_2022.values
})

# Calculating the difference
results_df['Difference'] = results_df['Votes_2022'] - results_df['Votes_2020']

# Displaying the DataFrame
display(results_df)
```

```python
# Saving the DataFrame to a CSV file
results_df.to_csv('Shifts_Since_2020.csv', index=False)
```

EXPLANATION: These are the changes in the number of registered voters for each political party between the 2020 and 2022 elections.
