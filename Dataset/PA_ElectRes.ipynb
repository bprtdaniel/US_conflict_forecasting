```python
import requests
import pandas as pd
from io import StringIO
```

```python
url = "https://www.dos.pa.gov/VotingElections/BEST/stats/Documents/ElectionReturns_2022_Primary_PrecinctReturns.txt"

#Parsing content of page, sending GET request 
response = requests.get(url)
response.raise_for_status()
```

```python
#Converting data into strinIO object 
data = StringIO(response.text)

#reading data into dataframe
PA_GenEl_Prec = pd.read_csv(data)

print(PA_GenEl_Prec.head())
```

```python
#Save as csv 
PA_GenEl_Prec.to_csv('PA_GenEl_Prec.csv', index=False)
```
