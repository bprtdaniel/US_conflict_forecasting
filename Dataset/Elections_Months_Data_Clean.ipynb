```python
# Loading file
county_zip_mapping = pd.read_csv('12_zip_ACS.csv')
```

```python
# Defining the election month starting November 1, 2022
election_date = pd.Timestamp('2022-11-01')

# Generating a date range from January 2022 to December 2022
date_range = pd.date_range(start='2022-01-01', end='2022-12-31', freq='MS')

# Creating a DataFrame with each county repeated 12 times (once for each month)
county_data = pd.concat([county_zip_mapping[['ZCTA', 'Name']].copy() for _ in range(12)], ignore_index=True)

# Adding the 'month' column with the date range repeated for each county
county_data['month'] = pd.concat([pd.Series(date_range)] * len(county_zip_mapping), ignore_index=True)

# Calculating months until and since the election
county_data['months_until_election'] = (election_date - county_data['month']).dt.days // 30
county_data['months_since_election'] = (county_data['month'] - election_date).dt.days // 30

# Adjusting the data so that only positive values are shown for months_until_election and months_since_election
county_data['months_until_election'] = county_data['months_until_election'].apply(lambda x: x if x >= 0 else None)
county_data['months_since_election'] = county_data['months_since_election'].apply(lambda x: x if x >= 0 else None)

# Converting the 'month' to format YYYY-MM
county_data['month'] = county_data['month'].dt.strftime('%Y-%m')

# Reordering the columns
county_data = county_data[['ZCTA', 'Name', 'month', 'months_until_election', 'months_since_election']]

# Saving the DataFrame to a CSV file named 'elections_month_data.csv'
county_data.to_csv('elections_month_data.csv', index=False)

print("Data has been saved to 'elections_month_data.csv'.")

```
