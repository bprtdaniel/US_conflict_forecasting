```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Loading the datasets
data_2022 = pd.read_csv('/content/PA_l2_2022stats_county.csv')

# Displaying the first 5 rows of the dataset
display(data_2022.head())
```

```python
# Printing the column names
print(data_2022.columns)
```

```python
# Checking for missing values
data_2022.dropna(inplace=True)

# Displaying the first 5 rows of the dataset
display(data_2022.head())
```

```python
# Extract only the needed columns
win_data_2022 = data_2022[['countyname', 'party_dem', 'party_rep']]

# Displaying the dataset
display(win_data_2022)
```

```python
# Extract only the needed columns
win_data_2022 = data_2022[['countyname', 'party_dem', 'party_rep']]

# Calculating the absolute win margin (difference in number of votes between parties)
win_data_2022['win_margin_abs'] = abs(win_data_2022['party_dem'] - win_data_2022['party_rep'])

# Calculating the win margin as a percentage
win_data_2022['win_margin_pct'] = (win_data_2022['win_margin_abs'] / (win_data_2022['party_dem'] + win_data_2022['party_rep'])) * 100

# Determining which party won
win_data_2022['winning_party'] = win_data_2022.apply(lambda x: 'Democrat' if x['party_dem'] > x['party_rep'] else 'Republican', axis=1)

# Displaying the updated DataFrame to see the new columns
display(win_data_2022)
```

```python
# Saving the DataFrame to CSV
win_data_2022.to_csv('elections_results_2022.csv', index=False)
print("Data has been saved to 'elections_results_2022.csv'.")
```
