```python
# Combines 'transposed' data from 'PA Demographic Data by ZCTA - Copy.xlsx' "Additional" sheet and 'acs_conflict_varv.csv' into the final output dataset 'acs_conflict_combined.csv'
# Uses '12_zip_ACS.csv' in the process to build ZIP structure
```

```python
import pandas as pd
import numpy as np
```

```python
zip = pd.read_csv('12_zip_ACS.csv')
acs_conflict = pd.read_csv('acs_conflict_varv.csv')
```

```python
transposed = pd.read_excel('PA Demographic Data by ZCTA - Copy.xlsx', sheet_name='Transposed Data')
```

```python
monthly_ZCTA_tr = pd.DataFrame()

months = pd.date_range(start='2022-01-01', end='2022-12-01', freq='MS').strftime('%Y-%m').tolist()

for index, row in transposed.iterrows():
    zip_code = row['ZCTA']
    zip_code_entries = pd.DataFrame({'ZCTA': [zip_code] * len(months), 'Month': months})
    monthly_ZCTA_tr = pd.concat([monthly_ZCTA_tr, zip_code_entries])
```

```python
transposed_zip_12 = pd.merge(monthly_ZCTA_tr, transposed, on = 'ZCTA', how = 'left')
```

```python
acs_last_month = pd.read_csv('acs_conflict_varv.csv')
```

```python
acs_conflict_combined = pd.merge(acs_last_month, transposed, on = 'ZCTA', how = 'inner')
```

```python
acs_conflict_combined.to_csv('acs_conflict_combined.csv', index=False)
```
