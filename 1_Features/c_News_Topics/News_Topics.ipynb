```python
#Dynamic search path configuration
import sys
import os

#Adjusting search path based on notebook's depth in directory --> iterates through directories until it finds directory name: US_conflict_forecasting
current_dir = os.getcwd()
while 'US_conflict_forecasting' not in os.path.basename(current_dir):
    current_dir = os.path.dirname(current_dir)

#Adjusting search path based on 'new' current directory
if current_dir not in sys.path:
    sys.path.insert(0, current_dir)

from project_setup import setup_project
project_root_dir = setup_project()

import pandas as pd
```

```python
#Loading Dataset 
News_topics_df = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/a_Input_Datasets/01-2023_conflictforecast_anyviolence_12.csv'))
```

```python
#Filtering relevant data only 
News_topics_22_USA = News_topics_df[(News_topics_df['isocode'] == 'USA') & (News_topics_df['year'] == 2022)]

#dropping unncessary variables
News_topics_22_USA = News_topics_22_USA.drop(['text_model', 
                                            'year',
                                            'isocode',
                                            'best_model', 
                                            'armedconf', 
                                            'anyviolence', 
                                            'armedconf_dp', 
                                            'anyviolence_dp', 
                                            'populationwb', 
                                            'lnbest_pc',
                                            'lnbest_pc_3', 
                                            'lnbest_pc_12', 
                                            'ons_anyviolence3', 
                                            'ons_armedconf3', 
                                            'ons_anyviolence12', 
                                            'ons_armedconf12'], axis=1)

#restructuring data column 
def format_month(month):
    return f'2022-{month:02}'

News_topics_22_USA['month'] = News_topics_22_USA['month'].apply(format_month)                     

News_topics_22_USA = News_topics_22_USA.reset_index(drop=True)
```

```python
#Checking how many unique ZIP Codes in final dataset
All_Conflict_Features = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets/All_Conflict_Features.csv'))

print(All_Conflict_Features['ZCTA'].nunique())
```

```python
#Replication country-level data 1833 times (nr of ZIP Codes) and resetting index 
#Can be merged into the final dataset on the index later 

nr_ZCTA = 1833

News_topics_22_USA_allZCTA = pd.concat([News_topics_22_USA] * nr_ZCTA, ignore_index=True)
```

```python
#Export to csv 
output_directory_2 = os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets')
output_filename_2 = "All_NewsTopic_Features.csv"
output_path_2 = os.path.join(output_directory_2, output_filename_2)

News_topics_22_USA_allZCTA.to_csv(output_path_2, index=False)
```
