```python
import pandas as pd

#Loading data
ACLED_df = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets/joined_ACLED_ZIP_22_PA.csv'))
```

```python
#check unique string values
ACLED_df['sub_event_type'].unique()
```

### Creating dummies for each sub event type

```python
#Defining sub event types
sub_event_types = [
    'Peaceful protest', 'Other', 'Protest with intervention',
    'Mob violence', 'Violent demonstration', 'Looting/property destruction'
]
```

```python
#Creating binary columns 
for event_type in sub_event_types:
    ACLED_df[event_type] = ACLED_df['sub_event_type'].apply(lambda x: 1 if x == event_type else 0)
```

```python
#Renaming columns
ACLED_df = ACLED_df.rename(columns={
    'Peaceful protest': 'set_peaceful_prot',
    'Other': 'set_other',
    'Protest with intervention': 'set_prot_int',
    'Mob violence': 'set_mob',
    'Violent demonstration': 'set_violent_prot',
    'Looting/property destruction': 'set_loot_destr'
})
```

```python
#Dropping all columns except the relevant Subevent features and ID
columns_to_keep = ['event_id_cnty', 'ZCTA5CE10', 'set_peaceful_prot', 'set_other', 'set_prot_int', 'set_mob', 'set_violent_prot', 'set_loot_destr']
ACLED_df_sparse = ACLED_df[columns_to_keep]
```

```python
#Saving as csv
output_directory_1 = os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets')
output_filename_1 = "ACLED_ZIP_Subevent_22.csv"
output_path_1 = os.path.join(output_directory_1, output_filename_1)

ACLED_df_sparse.to_csv(output_path_1, index=False)
```
