### Script to Join the Topics with ACLED Dataset
- Preparatation for aggregation of conflict events in ConflictDataset_cleaning.ipynb  
- Joining Topics data with Acled conflict data on Zip ID


Output: ACLED_ZIP_Actor_SET_22_PA


```python
#Dynamic search path configuration
import sys
import os

#Adjusting search path based on notebook's depth in directory --> iterates through directories until it finds directory name: US_conflict_forecasting
current_dir = os.getcwd()
while 'US_conflict_forecasting' not in os.path.basename(current_dir):
    current_dir = os.path.dirname(current_dir)

#Adjusting search path based on 'new' current directory
if current_dir not in sys.path:
    sys.path.insert(0, current_dir)

from project_setup import setup_project
project_root_dir = setup_project()
```

```python
import numpy as np
import pandas as pd
```

```python
#Loading ACLED Data 
ACLED_df = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets/joined_ACLED_ZIP_22_PA.csv'))

#Loading topics data 
Topics_df = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets/topics.csv'))
```

```python
#Joining on conflict ID
ACLED_Topics_Join = ACLED_df.merge(Topics_df, left_on='event_id_cnty', right_on='Event ID')

#Droping duplicate ID column
ACLED_Topics_Join = ACLED_Topics_Join.drop(columns=['Event ID'])
```

```python
#Dropping irrelevant columns
columns_to_keep = ['event_id_cnty', 'ZCTA5CE10', 'unions', 'environment', 'gun-violence', 'healthcare'] 
ACLED_Topics_Join_sparse = ACLED_Topics_Join[columns_to_keep]
```

```python
#Exporting to csv 
output_directory_1 = os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets')
output_filename_1 = "ACLED_ZIP_Topic.csv"
output_path_1 = os.path.join(output_directory_1, output_filename_1)

ACLED_Topics_Join_sparse.to_csv(output_path_1, index=False)
```
