```python
#Dynamic search path configuration
import sys
import os

#Adjusting search path based on notebook's depth in directory --> iterates through directories until it finds directory name: US_conflict_forecasting
current_dir = os.getcwd()
while 'US_conflict_forecasting' not in os.path.basename(current_dir):
    current_dir = os.path.dirname(current_dir)

#Adjusting search path based on 'new' current directory
if current_dir not in sys.path:
    sys.path.insert(0, current_dir)

from project_setup import setup_project
project_root_dir = setup_project()

import pandas as pd
```

```python
#Load data 
Election_results = pd.read_excel(os.path.join(project_root_dir, '2_Dataset/a_Input_Datasets/PA_ElectionResults.xlsx'))

Election_months = pd.read_csv(os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets/Election_Feature_Months.csv'))
```

```python
Election_results['admin2'] = Election_results['admin2'].astype(str)
Election_months['admin2'] = Election_months['admin2'].astype(str)

#merge datasets 
All_Election_Features = pd.merge(Election_months, Election_results, on='admin2', how='left')

#Ensuring that values are rounded
columns_to_round = ['Dem_VoteShare20_GenEl', 'Rep_VoteShare20_GenEl', 'DemWinMarginPP_20_GenEl','Shift_dem_20to22', 'Shift Rep_20to22' ]
All_Election_Features[columns_to_round] = All_Election_Features[columns_to_round].round(2)
```

```python
#save dataset
output_directory_1 = os.path.join(project_root_dir, '2_Dataset/b_Partial_Datasets')
output_filename_1 = "All_Election_Features.csv"
output_path_1 = os.path.join(output_directory_1, output_filename_1)

All_Election_Features.to_csv(output_path_1, index=False)
```
